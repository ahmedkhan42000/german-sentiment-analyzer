```python
from transformers import pipeline

# Load the German sentiment analysis model
# This downloads the model automatically the first time — may take 1-2 minutes
sentiment_model = pipeline(
    model="oliverguhr/german-sentiment-bert",
    task="text-classification"
)

print("Model loaded successfully!")
```

```python
# Test sentences in German
sentences = [
    "Dieses Produkt ist absolut fantastisch!",      # This product is absolutely fantastic
    "Das war eine schreckliche Erfahrung.",          # That was a terrible experience
    "Das Wetter heute ist okay.",                    # The weather today is okay
    "Ich liebe meine Arbeit als KI-Ingenieur!",     # I love my work as AI engineer
    "Der Service war sehr enttäuschend."             # The service was very disappointing
]

# Run each sentence through the model
for sentence in sentences:
    result = sentiment_model(sentence)
    label = result[0]['label']
    score = round(result[0]['score'] * 100, 2)
    print(f"Text: {sentence}")
    print(f"Sentiment: {label} | Confidence: {score}%")
    print("-" * 60)
```

```python
import requests
import os
import pandas as pd
from dotenv import load_dotenv

# Load token
load_dotenv()
token = os.getenv("HF_TOKEN")

# API settings
API_URL = "https://api-inference.huggingface.co/models/oliverguhr/german-sentiment-bert"
headers = {"Authorization": f"Bearer {token}"}

def analyze_sentiment(text):
    response = requests.post(API_URL, headers=headers, json={"inputs": text})
    result = response.json()
    label = result[0][0]['label']
    score = round(result[0][0]['score'] * 100, 2)
    return label, score

# A bigger set of real German sentences
test_sentences = [
    "Dieses Produkt ist absolut fantastisch!",
    "Das war eine schreckliche Erfahrung.",
    "Das Wetter heute ist okay.",
    "Ich liebe meine Arbeit als KI-Ingenieur!",
    "Der Service war sehr enttäuschend.",
    "Die Lieferung kam pünktlich an.",
    "Ich bin sehr zufrieden mit dem Kauf.",
    "Das Essen war komplett ungenießbar.",
    "Ganz normale Erfahrung, nichts Besonderes.",
    "Absolut empfehlenswert, top Qualität!"
]

# Run all sentences through the model
results = []
for sentence in test_sentences:
    label, score = analyze_sentiment(sentence)
    results.append({
        "text": sentence,
        "sentiment": label,
        "confidence": score
    })
    print(f"{label} ({score}%) — {sentence}")

# Save to CSV
df = pd.DataFrame(results)
df.to_csv("sentiment_results.csv", index=False)
print("\nResults saved to sentiment_results.csv")
```
